# Run pipeline (01 → 04)

Runs statistics, panel plotting, main figure, and split-half supplement using
`data/analysis_input.csv`.

```python
from pathlib import Path

import nbformat
from nbconvert.preprocessors import ExecutePreprocessor

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
NB = PROJECT_ROOT / "notebooks"
DATA = PROJECT_ROOT / "data" / "analysis_input.csv"
if not DATA.is_file():
    raise FileNotFoundError(f"Missing analysis input: {DATA}")

steps = [
"01_statistics.ipynb",
"02_panel_plotting.ipynb",
"03_main_figure.ipynb",
"04_split_half_supplement.ipynb",
]
ep = ExecutePreprocessor(timeout=1800, kernel_name="python3")
for s in steps:
    print(f"\n=== {s} ===")
    with open(NB / s, encoding="utf-8") as f:
        nb = nbformat.read(f, as_version=4)
        ep.preprocess(nb, {"metadata": {"path": str(NB)}})
        with open(NB / s, "w", encoding="utf-8") as f:
            nbformat.write(nb, f)

print("\nDone. See figures/ and results/split_half/.")
```
