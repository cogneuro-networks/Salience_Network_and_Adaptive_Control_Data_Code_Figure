# Run example pipeline (01 → 02 → 03 → 04)

Runs notebooks 01–04 with `data_mode: example` and `example_data/`.


```python
import yaml
from pathlib import Path
import nbformat
from nbconvert.preprocessors import ExecutePreprocessor

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
NB = PROJECT_ROOT / "notebooks"
cfg = yaml.safe_load(open(PROJECT_ROOT / "config.yaml", encoding="utf-8"))
cfg["data_mode"] = "example"
yaml.dump(cfg, open(PROJECT_ROOT / "config.yaml", "w", encoding="utf-8"), allow_unicode=True, sort_keys=False)

steps = ["01_roi_and_parcel_selection.ipynb","02_hmm_decode_and_state_fc.ipynb",
"03_weighted_average_fc.ipynb","04_compute_functional_gradients.ipynb"]
ep = ExecutePreprocessor(timeout=900, kernel_name="python3")
for s in steps:
    print(f"\n=== {s} ===")
    with open(NB/s, encoding="utf-8") as f: nb = nbformat.read(f, as_version=4)
    ep.preprocess(nb, {"metadata": {"path": str(NB)}})
    with open(NB/s, "w", encoding="utf-8") as f: nbformat.write(nb, f)
    print("\nDone. See results/example/")
```
